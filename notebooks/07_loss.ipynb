```python
import torch
import torch.nn as nn
from src.model import Transformer

vocab_size = 8000
pad_id = 0
model = Transformer(vocab_size, pad_id=pad_id)
loss_fn = nn.CrossEntropyLoss(ignore_index=pad_id, label_smoothing=0.1)

src = torch.randint(1, vocab_size, (2, 10))
tgt_input = torch.randint(1, vocab_size, (2, 6))
tgt_label = torch.randint(1, vocab_size, (2, 6))

output = model(src, tgt_input)
loss = loss_fn(output.view(-1, vocab_size), tgt_label.view(-1))
print(loss.item())
```
